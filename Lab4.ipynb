```python
def tegsh_uu(n):
    if n % 2 == 0: 
        return "Тэгш тоо" 
    else: 
        return "Сондгой тоо" 

too = int(input("Тоо оруулна уу: ")) 
print(tegsh_uu(too))
```

```python
def find_max(a, b, c):
    if a >= b and a >= c:
        return a
    elif b >= a and b >= c:
        return b
    else:
        return c

# Хэрэглэгчээс 3 тоог авч шалгах
x = int(input("1-р тоо: "))
y = int(input("2-р тоо: "))
z = int(input("3-р тоо: "))
print("Хамгийн их нь:", find_max(x, y, z))
```
