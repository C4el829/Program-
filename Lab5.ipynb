```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

add_result = a + b
print("a + b = ", add_result)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])
b = np.array([5, 6, 7, 8])

mul_result = a * b
print("a * b = ", mul_result)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])

sq_result = a ** 2
print("a-ийн квадрат = ", sq_result)
```

```python
import numpy as np

a = np.array([1, 2, 3, 4])

mean_a = np.mean(a)
sum_a = np.sum(a)
max_a = np.max(a)

print(f"a массивын дундаж: {mean_a}")
print(f"a массивын нийлбэр: {sum_a}")
print(f"a массивын хамгийн их утга: {max_a}")
```
