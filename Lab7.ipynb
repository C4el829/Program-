```python
import pandas as pd

s = pd.Series(
    [45000, 32000, 52000, 41000],
    index=["Tokyo", "London", "New York", "Seoul"]
)

# Индексээр эрэмбэлэх
sorted_s = s.sort_index()

print(sorted_s)
```

```python
import pandas as pd

scores = pd.Series(
    [88, 95, 70, 100, 82],
    index=["Bat", "Bold", "Saraa", "Temuujin", "Anu"]
)

# Өсөхөөр эрэмбэлэх
ascending_scores = scores.sort_values()

# Буурахаар эрэмбэлэх
descending_scores = scores.sort_values(ascending=False)

print("Өсөхөөр:")
print(ascending_scores)

print("Буурахаар:")
print(descending_scores)
```

```python
import pandas as pd

countries = pd.Series(
    ["Mongolia", "Japan", "China", "Korea", "Germany"]
)

# Japan, Germany байгаа эсэх
check = countries.isin(["Japan", "Germany"])

# Зөвхөн эдгээр улсууд
filtered = countries[check]

print("Шалгалт:")
print(check)

print("Шүүсэн утгууд:")
print(filtered)
```

```python
import pandas as pd

population = pd.Series(
    [3.4, 126, 83],
    index=["Mongolia", "Japan", "Germany"]
)

# NumPy array
values_array = population.values

# List болгох
values_list = list(values_array)

print("NumPy array:")
print(values_array)

print("List:")
print(values_list)
```

```python
import pandas as pd

cities = pd.Series(
    [1.5, 9.7, 8.9],
    index=["Ulaanbaatar", "Tokyo", "New York"]
)

# Index авах
city_indexes = cities.index

# List болгох
city_list = list(city_indexes)

print("Index:")
print(city_indexes)

print("List:")
print(city_list)
```

```python
import pandas as pd

prices = pd.Series(
    [2500, 3000, 1800],
    index=["Bread", "Milk", "Rice"]
)

# items ашиглах
for item, price in prices.items():
    print(f"{item} price = {price}")
```

```python
import pandas as pd

countries = pd.Series(
    ["Paris", "Berlin", "Rome", "Madrid", "Lisbon"],
    index=["France", "Germany", "Italy", "Spain", "Portugal"]
)

# Label slicing
print("Label slicing:")
print(countries["Germany":"Spain"])

# Numeric slicing
print("Numeric slicing:")
print(countries[1:4])
```
