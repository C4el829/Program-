```python
import pandas as pd

# Өгөгдөл
data = {
    "Нэр": ["Бат", "Сараа", "Тэмүүжин"],
    "Нас": [20, 22, 19],
    "Дүн": [85, 90, 78]
}

df = pd.DataFrame(data)

# 1. "Тэнцсэн" багана нэмэх
df["Тэнцсэн"] = df["Дүн"] > 60

print("=== Шинэ баганатай DataFrame ===")
print(df)

# 2. 80-аас дээш дүнтэй сурагчид
print("\n=== 80-аас дээш дүнтэй ===")
print(df[df["Дүн"] > 80])

# Нас нь 20-оос их
print("\n=== Нас нь 20-оос их ===")
print(df[df["Нас"] > 20])

# 3. Дүнгийн дундаж
average_score = df["Дүн"].mean()

print("\n=== Дундаж дүн ===")
print(average_score)
```

```python
import pandas as pd

# Өгөгдөл
data = {
    "Нэр": ["Бат", "Сараа", "Тэмүүжин", "Ану", "Болд"],
    "Анги": ["A", "A", "B", "B", "A"],
    "Дүн": [85, 90, 78, 88, 92]
}

df = pd.DataFrame(data)

print("=== DataFrame ===")
print(df)

# 1. Анги бүрийн дундаж дүн
print("\n=== Анги бүрийн дундаж дүн ===")
print(df.groupby("Анги")["Дүн"].mean())

# 2. Анги бүрийн хамгийн өндөр дүн
print("\n=== Анги бүрийн хамгийн өндөр дүн ===")
print(df.groupby("Анги")["Дүн"].max())

# 3. Дүнгээр буурахаар эрэмбэлэх
print("\n=== Дүнгээр эрэмбэлсэн ===")
print(df.sort_values(by="Дүн", ascending=False))
```
